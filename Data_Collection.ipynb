```python
# Collecting Data

import wikipediaapi

topics = [
    "Anthropic",
    "Jev (AI model)",
    "Grok (chatbot)",
    "GPT-6",
    "Claude Mythos"
]


for topic in topics:
    wiki = wikipediaapi.Wikipedia(user_agent = "AI RAG v1.0 (personal project)", language = "en")
    page = wiki.page(topic)
    if page.exists():
        with open(f"Data/{topic}.txt", "w", encoding= "utf-8") as f:
            f.write(page.text)
            print(f"{topic} Added!")
    else:
        print(f"{topic} not found")

```

```python
# A Function to Create the Database

from pathlib import Path
import chromadb

def create_database(path_dir):

    client = chromadb.PersistentClient(path= "rag_db")
    collection = client.get_or_create_collection("AI-Information")
    main_dir = path_dir
    for file_path in Path(main_dir).glob("*.txt"):
        with open(file_path, "r", encoding= "utf-8") as f:
            topic_name = file_path.stem
            print(f"Starting Topic : {topic_name}")
            lines = f.readlines()
            chunks = []
            for line in lines:
                if len(line) > 40:
                    chunks.append(line)

            print(f"{len(chunks)} Chunks Created for Topic : {topic_name}")

            for i, chunk in enumerate(chunks):
                collection.add(
                    ids= f"{topic_name}_{i}",
                    documents= chunk,
                    metadatas={"source" : f"{topic_name}_{i}", "type" : "text"}
                    
                )
            print(f"{topic_name} added to DB !")
            


```

```python
# Lets Run the function on Path "Data"

create_database(path_dir= "Data")
```

```python
# Query The DataBase

import chromadb

client = chromadb.PersistentClient(path= "rag_db")
collection = client.get_collection("AI-Information")

result = collection.query(
    query_texts = ["What is Claude Mythos"],
    n_results= 2
)

print(result['documents'][0][0])
```
